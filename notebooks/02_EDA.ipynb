# Missing Values

```python
missing = application_train.isnull().sum()
```

```python
missing = missing[missing > 0]

missing = (
    missing
    .sort_values(ascending=False)
    .to_frame("Missing")
)

missing["Percent"] = (
    missing["Missing"] /
    len(application_train)
) * 100

missing.head(30)
```

```python
plt.figure(figsize=(10,8))

missing["Percent"].head(30).plot(kind="barh")

plt.title("Top Missing Columns")

plt.show()
```
